## **List, Tuple, and Array**

### **List**

*   **Definition:** A built-in, ordered, and mutable collection that can store  heterogeneous data types.
*   **Syntax:** Defined using [].
*   **Mutability:** Elements can be added, removed, or changed after creation.




```python
a = [1,2,'apple']
b = a
b[0]=3
b.append('orange')
print(a)
```

### **Tuple**

*   **Definition:** A built-in, ordered, and immutable collection that can store heterogeneous data types.
*   **Syntax:** Defined using ().
*   **Mutability:** Elements cannot be changed, added, or removed after creation.



```python
my_tuple = (1, "hello", 3.14)
print(my_tuple[1])
#my_tuple[1]="hi"
```

### **Array (NumPy Array)**

*   **Definition:** A data structure provided by the NumPy library, optimized for numerical operations and storing homogeneous data (all elements of the same data type).
*   **Syntax:** Created using numpy.array().
*   **Mutability:** Mutable, allowing modification of elements.
*   **Efficiency:** More memory-efficient and faster for numerical computations compared to Python lists.



```python
import numpy as np
my_array = np.array([1, 2, 3, 4])
my_array[0] = 5 # Modifies the array
print(my_array)
```

*   **Syntax:** Created using {}.

**The code below uses pymssql makes connection to the Cloud SQL database and executes a query. The query output is stored in a pandas object**

```python
from google.colab import auth
import pandas as pd
import sqlalchemy
import matplotlib.pyplot as plt
from google.cloud.sql.connector import Connector

# Authenticate Colab user (must have Cloud SQL Client role on the GCP project)
auth.authenticate_user()

# Configuration
INSTANCE_CONNECTION_NAME ='bis638-433901:us-east4:free-trial-first-project' #'mimic-iv-430316:us-east4:mytestproject'
DB_USER = "kei"         #ian or kei # or "root" / custom user
DB_PASS = "Cloud$4SQL" #cloud$sql or Cloud$4SQL
DB_NAME = "MIMICIV"            # your database name

# Initialize Connector
connector = Connector()

# Helper function to generate connections
def getconn():
    conn = connector.connect(
        INSTANCE_CONNECTION_NAME,
        "pymysql",              # Driver: 'pg8000' for Postgres, 'pymysql' for MySQL
        user=DB_USER,
        password=DB_PASS,
        db=DB_NAME
    )
    return conn

engine = sqlalchemy.create_engine(
    "mysql+pymysql://",      # Changed from 'postgresql+pg8000://'
    creator=getconn,
)

query = """
SELECT race, count(*) as ct FROM Admissions group by race
"""

# Run query and fetch into DataFrame
with engine.connect() as connection:
    df = pd.read_sql(sqlalchemy.text(query), con=connection)

# Display results
print(df)
```

**The code below use matplotlib to create a piechart and a barchart for displaying the query output**

```python
import matplotlib.pyplot as plt

races=df['race'].values.tolist()
data=df['ct'].values.tolist()
total=sum(data)


plt.pie(data, autopct = lambda p: '{:.0f}%'.format(p * total / 100),labels = races, textprops={'fontsize': 6})
# show plot
plt.show()
plt.bar(races, data, color ='maroon',
        width = 0.4)

plt.xlabel("Race")
plt.xticks(rotation='vertical') #'vertical'
plt.ylabel("No. of Subjects")
plt.title("Subjects in Different Races")
plt.show()
```

### Create and execute a query that returns the number of unique subjects for each fluid type

```python
sql_query = '''select count(distinct a.subject_id) as ct,b.fluid
from `bis638-433901.test.labevents` a
inner join `bis638-433901.test.d_labitems` b on a.itemid=b.itemid
group by b.fluid
order by ct
'''
df = client.query(sql_query).to_dataframe()
print (df)
```

### Plotting the query results

```python
import matplotlib.pyplot as plt

fluids=df['fluid'].values.tolist()
data=df['ct'].values.tolist()
total=sum(data)


plt.pie(data, autopct = lambda p: '{:.0f}%'.format(p * total / 100),labels = fluids, textprops={'fontsize': 6})
# show plot
plt.show()
plt.bar(fluids, data, color ='maroon',
        width = 0.4)

plt.xlabel("Fluid")
plt.xticks(rotation='vertical') #'vertical'
plt.ylabel("No. of Subjects")
plt.title("Subjects in Different Fluid Types")
plt.show()
```
